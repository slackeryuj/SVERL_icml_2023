```python
import numpy as np
import gymnasium as gym
from gymnasium import spaces
from scipy.optimize import minimize

class PortfolioEnv(gym.Env):
    """
    Monthly Mean-CVaR portfolio environment.

    Reward (paid once per month at rebalance boundary):
        reward_t = month_ret_t  -  risk_coefficient * CVaR( monthly_ret_history )
                     - lambda_turnover * turnover_t
                     - lambda_hhi * HHI(weights_t)
                     - transaction_cost_rate * turnover_t

    Key behaviors:
    - Rebalance every `rebalance_period` days using action -> weights mapping
      that supports baseline-relative or delta-to-current semantics.
    - Between rebalances, weights drift self-financing.
    - Observations are a flat vector of the last `lookback_period` days of features.
    """

    metadata = {"render_modes": []}

    def __init__(
        self,
        data,
        etf_list,
        reward_type="mean_cvar",
        risk_coefficient=0.5,
        rebalance_period=21,
        lookback_period=21,
        weight_bounds=(0.0, 1.0),            # (min_weight, max_weight)
        desired_long=1.0,                    # total long budget (e.g., 1.2 for 120/20)
        desired_short=0.0,                   # total short budget (e.g., 0.2 for 120/20)
        use_baseline=False,
        baseline_fn=None,                    # fn(date) -> weight vector
        transaction_cost_rate=0.0,
        lambda_turnover=0.001,
        lambda_hhi=0.1,
        lambda_deviation=0.0,
        # Enhancements / numerics:
        alpha=0.05,                          # CVaR tail level
        cvar_warmup_months=12,               # months before CVaR kicks in
        exclude_price_levels=True,           # drop columns starting with "Price_"
        allow_actual_returns_in_obs=False,
        action_temperature=1.0,              # scale for tanh squashing
        daily_return_clip=None,            
        shap_align_coef: float = 0.0
    ):
        super().__init__()

        # ---------- Data / config ----------
        self.data = data.reset_index(drop=True)
        self.etf_list = list(etf_list)
        self.reward_type = reward_type
        self.risk_coefficient = float(risk_coefficient)
        self.rebalance_period = int(rebalance_period)
        self.lookback_period = int(lookback_period)
        self.weight_bounds = tuple(weight_bounds)
        self.desired_long = float(desired_long)
        self.desired_short = float(desired_short)
        self.use_baseline = bool(use_baseline)
        self.baseline_fn = baseline_fn
        self.transaction_cost_rate = float(transaction_cost_rate)
        self.lambda_turnover = float(lambda_turnover)
        self.lambda_hhi = float(lambda_hhi)
        self.lambda_deviation = float(lambda_deviation)   # <--- NEW
        self.alpha = float(alpha)
        self.cvar_warmup_months = int(cvar_warmup_months)
        self.exclude_price_levels = bool(exclude_price_levels)
        self.allow_actual_returns_in_obs = bool(allow_actual_returns_in_obs)
        self.action_temperature = float(action_temperature)
        self.daily_return_clip = daily_return_clip if daily_return_clip is None else float(daily_return_clip)
        self._opt_baseline_cache = {}
        self._date_to_idx = {d: i for i, d in enumerate(self.data["Date"])}
        self.shap_align_coef = float(shap_align_coef)
        self.shap_dir_cols  = [f"SHAP_DIR_{e}_CSZ"  for e in self.etf_list]
        self.shap_conf_cols = [f"SHAP_CONF_{e}_CSZ" for e in self.etf_list]
        self.shap_dir_idx   = np.array([self.data.columns.get_loc(c)  if c in self.data.columns else -1 for c in self.shap_dir_cols])
        self.shap_conf_idx  = np.array([self.data.columns.get_loc(c) if c in self.data.columns else -1 for c in self.shap_conf_cols])
        self._has_shap = np.all(self.shap_dir_idx >= 0) and np.all(self.shap_conf_idx >= 0)


        # ---------- Validate required columns ----------
        for etf in self.etf_list:
            col = f"Actual_Return_{etf}"
            if col not in self.data.columns:
                raise ValueError(f"Missing required column '{col}' in data.")
        if "Date" not in self.data.columns:
            raise ValueError("Data must contain a 'Date' column.")

        # ---------- Observation features ----------
        # Exclude date & realized returns; optionally exclude raw prices
        def keep_feature(c: str) -> bool:
            if c == "Date":
                return False
            # only drop realized returns if flag is False
            if (not self.allow_actual_returns_in_obs) and c.startswith("Actual_Return_"):
                return False
            # optionally drop raw prices for stability
            if self.exclude_price_levels and c.startswith("Price_"):
                return False
            return True

        self.feature_cols = [c for c in self.data.columns if keep_feature(c)]
        self.num_features_per_day = len(self.feature_cols)

        self.action_space = spaces.Box(
            low=-1.0, high=1.0, shape=(len(self.etf_list),), dtype=np.float32
        )
        self.observation_space = spaces.Box(
            low=-np.inf, high=np.inf,
            shape=(self.num_features_per_day * self.lookback_period,),
            dtype=np.float32
        )

        # ---------- State ----------
        self.current_step = self.lookback_period
        self.cumulative_wealth = 1.0
        self.current_weights = np.ones(len(self.etf_list)) / len(self.etf_list)

        # Monthly bookkeeping
        self.daily_port_ret_buffer = []   # daily portfolio returns in current month
        self.monthly_ret_history = []     # realized monthly returns
        self._last_turnover = 0.0
        self._last_deviation = 0.0

    # ------------------------ Helpers ------------------------

    def _get_obs(self):
        start = self.current_step - self.lookback_period
        end = self.current_step
        obs_window = self.data.iloc[start:end]
        obs_values = obs_window[self.feature_cols].to_numpy(dtype=np.float32).reshape(-1)
        if not np.all(np.isfinite(obs_values)):
            obs_values = np.nan_to_num(obs_values, nan=0.0, posinf=0.0, neginf=0.0)
        return obs_values

    def _self_financing_drift(self, returns_today: np.ndarray) -> None:
        """Update weights passively between rebalances (long/short compatible)."""
        gross = self.current_weights * (1.0 + returns_today)
        denom = 1.0 + float(np.dot(self.current_weights, returns_today))
        self.current_weights = gross / max(denom, 1e-12)

    def _baseline_weights(self, date):
        """Fetch and sanitize baseline weights (non-negative & normalized)."""
        if self.baseline_fn is None:
            return np.ones(len(self.etf_list)) / len(self.etf_list)
        w = np.asarray(self.baseline_fn(date), dtype=float)
        if not np.all(np.isfinite(w)) or w.shape[0] != len(self.etf_list):
            w = np.ones(len(self.etf_list)) / len(self.etf_list)
        # For safety, clip negatives and renormalize baseline
        w = np.clip(w, 0.0, None)
        s = w.sum()
        return w / s if s > 0 else np.ones(len(self.etf_list)) / len(self.etf_list)
    
    # def _baseline_weights(self, date):
    #     """
    #     Baseline = Max-CVaR-Sharpe (mean / CVaR) optimizer over past 252 trading days,
    #     using monthly compounding over 'rebalance_period' day windows.
    #     Long-only, fully-invested. Cached by date for speed.
    # 
    #     Falls back to equal-weight if not enough history or solver fails.
    #     """
    #     # If caller provided an external baseline_fn, honor that first.
    #     if callable(self.baseline_fn):
    #         w = np.asarray(self.baseline_fn(date), dtype=float)
    #         if not np.all(np.isfinite(w)) or w.shape[0] != len(self.etf_list):
    #             w = np.ones(len(self.etf_list)) / len(self.etf_list)
    #         w = np.clip(w, 0.0, None)
    #         s = w.sum()
    #         return w / s if s > 0 else np.ones(len(self.etf_list)) / len(self.etf_list)
    # 
    #     # Use cached result if we already solved this date.
    #     if hasattr(self, "_opt_baseline_cache") and date in self._opt_baseline_cache:
    #         return self._opt_baseline_cache[date].copy()
    # 
    #     # ------------------ Settings for this baseline ------------------
    #     lookback    = 252
    #     horizon     = int(getattr(self, "rebalance_period", 21))   # monthly window length
    #     alpha       = float(getattr(self, "alpha", 0.05))          # CVaR tail (use env's alpha)
    #     cvar_floor  = 1e-6                                         # avoids div-by-zero in ratio
    # 
    #     # ------------------ Gather lookback returns ---------------------
    #     try:
    #         idx = self._date_to_idx[date]   # prebuilt in __init__
    #     except Exception:
    #         # Fallback: locate by scan (robust)
    #         idx = int(self.data.index[self.data["Date"] == date][0])
    # 
    #     # Need at least 252 days ending the day before 'date'
    #     if idx < lookback or horizon > lookback:
    #         w = np.ones(len(self.etf_list)) / len(self.etf_list)
    #         if hasattr(self, "_opt_baseline_cache"):
    #             self._opt_baseline_cache[date] = w
    #         return w.copy()
    # 
    #     cols = [f"Actual_Return_{etf}" for etf in self.etf_list]
    #     X = self.data.loc[idx - lookback: idx - 1, cols].to_numpy(dtype=float)   # [lookback, N]
    #     if not np.all(np.isfinite(X)):
    #         X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)
    # 
    #     # All rolling 'horizon' windows inside the lookback: shape [S, horizon, N]
    #     S = lookback - horizon + 1
    #     windows = np.stack([X[i:i + horizon, :] for i in range(S)], axis=0)
    # 
    #     # ------------------ Objective: maximize mean / CVaR ------------------
    #     def cvar_alpha(returns_1d: np.ndarray) -> float:
    #         m = returns_1d.shape[0]
    #         k = max(1, int(np.ceil(alpha * m)))
    #         worst = np.sort(returns_1d)[:k]
    #         return max(0.0, -float(np.mean(worst)))  # positive loss measure
    # 
    #     def objective(w: np.ndarray) -> float:
    #         # monthly compounded return for each scenario window
    #         daily = windows @ w                         # [S, horizon]
    #         month = np.prod(1.0 + daily, axis=1) - 1.0  # [S]
    #         mean_month = float(np.mean(month))
    #         loss_cvar  = cvar_alpha(month)
    #         denom = max(loss_cvar, cvar_floor)
    #         ratio = mean_month / denom                 # Max-CVaR-Sharpe
    #         return -ratio                              # minimize negative to maximize ratio
    # 
    #     # ------------------ Solve (long-only, sum=1) --------------------
    #     from scipy.optimize import minimize  # local import is fine; or keep at top-level
    # 
    #     n = len(self.etf_list)
    #     bounds = [(0.0, 1.0)] * n
    #     cons   = [{"type": "eq", "fun": lambda w: np.sum(w) - 1.0}]
    #     initials = [np.ones(n)/n] + [np.random.dirichlet(np.ones(n)) for _ in range(2)]
    # 
    #     best_val, best_w = float("inf"), None
    #     for w0 in initials:
    #         res = minimize(objective, w0, method="SLSQP",
    #                        bounds=bounds, constraints=cons,
    #                        options={"maxiter": 160, "ftol": 1e-8})
    #         if res.success and res.fun < best_val:
    #             best_val, best_w = res.fun, res.x
    # 
    #     if best_w is None or not np.all(np.isfinite(best_w)):
    #         w = np.ones(n) / n
    #     else:
    #         w = np.clip(best_w, 0.0, None)
    #         s = w.sum()
    #         w = w / s if s > 0 else (np.ones(n) / n)
    # 
    #     # Cache and return
    #     if hasattr(self, "_opt_baseline_cache"):
    #         self._opt_baseline_cache[date] = w.copy()
    #     return w.copy()


    def _map_action_to_weights(self, action: np.ndarray, date) -> np.ndarray:
        """
        Map agent action to portfolio weights with long/short budgets and bounds.
        - Supports baseline-relative (if self.use_baseline) or delta-to-current.
        - Enforces desired long/short budgets after clipping to weight_bounds.
        """
        a = np.asarray(action, dtype=float)
        if not np.all(np.isfinite(a)):
            a = np.zeros_like(self.current_weights)

        # Smoothly squash actions to (-1, 1)
        temp = max(self.action_temperature, 1e-6)
        a = np.tanh(a / temp)

        if self.use_baseline:
            base = self._baseline_weights(date)
            raw = base * (1.0 + a)            # relative tilt around baseline
        else:
            raw = self.current_weights + a     # delta to current weights

        # Split long/short parts
        long_w = np.clip(raw, 0.0, None)
        short_w = np.clip(-raw, 0.0, None)

        has_long = long_w.sum() > 0
        has_short = short_w.sum() > 0
        n = len(raw)

        if has_long and has_short:
            norm_long = self.desired_long * long_w / long_w.sum()
            norm_short = self.desired_short * short_w / short_w.sum()
        elif has_long and not has_short:
            norm_long = long_w / long_w.sum()         # 100% long
            norm_short = np.zeros(n)
        elif not has_long and has_short:
            # Fallback: equal-weight long-only (avoid pathological all-short)
            norm_long = np.ones(n) / n
            norm_short = np.zeros(n)
        else:
            # All zeros -> equal-weight long-only
            norm_long = np.ones(n) / n
            norm_short = np.zeros(n)

        combined = norm_long - norm_short

        # Clip to hard bounds, then renormalize long/short budgets
        lo, hi = self.weight_bounds
        clipped = np.clip(combined, lo, hi)

        long_c = np.clip(clipped, 0.0, None)
        short_c = np.clip(-clipped, 0.0, None)

        if long_c.sum() > 0 and short_c.sum() > 0:
            final_long = self.desired_long * long_c / long_c.sum()
            final_short = self.desired_short * short_c / short_c.sum()
        elif long_c.sum() > 0:
            final_long = long_c / long_c.sum()
            final_short = np.zeros(n)
        else:
            final_long = np.ones(n) / n
            final_short = np.zeros(n)

        return final_long - final_short

    # ------------------------ Gym API ------------------------

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        if seed is not None:
            np.random.seed(seed)
        self.current_step = self.lookback_period
        self.cumulative_wealth = 1.0
        self.current_weights = np.ones(len(self.etf_list)) / len(self.etf_list)
        self.daily_port_ret_buffer = []
        self.monthly_ret_history = []
        self._last_turnover = 0.0
        obs = self._get_obs()
        return obs, {}

    def step(self, action):
        next_step = self.current_step + 1
        reward = 0.0
        terminated = False

        # 1) Rebalance day: map action -> new weights, compute turnover
        if self.current_step % self.rebalance_period == 0:
            prev_w = self.current_weights.copy()
            current_date = self.data.loc[self.current_step, "Date"]
            new_w = self._map_action_to_weights(action, current_date)
            # L1 turnover
            self._last_turnover = float(np.sum(np.abs(new_w - prev_w)))
            # L1 deviation vs baseline (cached inside _baseline_weights)
            base_w = self._baseline_weights(current_date)
            self._last_deviation = float(np.sum(np.abs(new_w - base_w)))
            self.current_weights = new_w

        # 2) Realize today's return (t -> t+1)
        if next_step >= len(self.data):
            # No next day to realize
            terminated = True
        else:
            asset_returns = np.array(
                [self.data.loc[next_step, f"Actual_Return_{etf}"] for etf in self.etf_list],
                dtype=float
            )
            # Optional daily return clipping (rarely needed if features are clean)
            port_ret = float(np.dot(self.current_weights, asset_returns))
            if self.daily_return_clip is not None:
                port_ret = float(np.clip(port_ret, -self.daily_return_clip, self.daily_return_clip))
            if not np.isfinite(port_ret):
                port_ret = 0.0

            self.cumulative_wealth *= (1.0 + port_ret)
            self.daily_port_ret_buffer.append(port_ret)

            # Passive drift on non-rebalance days
            if self.current_step % self.rebalance_period != 0:
                self._self_financing_drift(asset_returns)

            # 3) Pay monthly reward at boundary (or at the end)
            end_of_month = (next_step % self.rebalance_period == 0)
            last_step = (next_step == len(self.data) - 1)
            if end_of_month or last_step:
                month_ret = np.prod([1.0 + r for r in self.daily_port_ret_buffer]) - 1.0
                self.daily_port_ret_buffer = []

                # CVaR on time-series of monthly returns (excluding current month)
                cvar_loss = 0.0
                if len(self.monthly_ret_history) >= self.cvar_warmup_months:
                    arr = np.array(self.monthly_ret_history, dtype=float)
                    var = np.percentile(arr, 100.0 * self.alpha)
                    tail = arr[arr <= var]
                    if tail.size > 0:
                        cvar_loss = -float(np.mean(tail))  # positive loss

                # Base reward
                if self.reward_type == "mean_cvar":
                    base_reward = month_ret - self.risk_coefficient * cvar_loss
                elif self.reward_type == "cumulative_return":
                    base_reward = month_ret
                elif self.reward_type == "log_wealth":
                    base_reward = float(np.log(max(1.0 + month_ret, 1e-12)))
                elif self.reward_type == "mean_var":
                    base_reward = month_ret
                elif self.reward_type == "cvar_sharpe":
                    denom = max(cvar_loss, 1e-6)
                    base_reward = month_ret / denom
                else:
                    raise ValueError(f"Invalid reward type: {self.reward_type}")
                
                # Start from base
                reward = base_reward
                
                # SHAP alignment bonus
                if self.shap_align_coef > 0.0 and self._has_shap:
                    row = self.data.iloc[next_step].to_numpy()
                    shap_dir  = row[self.shap_dir_idx]
                    shap_conf = row[self.shap_conf_idx]
                    shap_signal = shap_dir * np.maximum(shap_conf, 0.0)
                    shap_signal = np.nan_to_num(shap_signal)
                    n = np.linalg.norm(shap_signal)
                    if n > 0:
                        shap_unit = shap_signal / n
                        w = np.asarray(self.current_weights, float)
                        w = w / (np.linalg.norm(w) + 1e-8)
                        align = float(np.clip(np.dot(w, shap_unit), -1.0, 1.0))
                        reward += self.shap_align_coef * align
                
                # Penalties/costs — subtract from *current* reward (do not reset)
                hhi_val = float(np.sum(np.square(np.abs(self.current_weights))))
                penalty = (
                    self.lambda_hhi * hhi_val
                    + self.lambda_turnover * self._last_turnover
                    + self.lambda_deviation * self._last_deviation
                )
                costs = self.transaction_cost_rate * self._last_turnover
                reward = reward - penalty - costs

                # Update monthly history AFTER computing reward
                self.monthly_ret_history.append(month_ret)

            terminated = next_step >= (len(self.data) - 1)

        self.current_step += 1
        obs = self._get_obs()
        if not np.all(np.isfinite(obs)):
            obs = np.nan_to_num(obs, nan=0.0, posinf=0.0, neginf=0.0)

        info = {
            "date": self.data.loc[min(self.current_step, len(self.data)-1), "Date"],
            "weights": self.current_weights.copy(),
            "cumulative_wealth": float(self.cumulative_wealth),
            "last_turnover": float(self._last_turnover),
            "num_months": len(self.monthly_ret_history),
            "last_deviation": float(self._last_deviation),
        }
        return obs, float(reward), bool(terminated), False, info

```

```python
# ===============================
# Stage 2 — Upgraded driver
# ===============================
import os
import json
import time
import gc
import random
from dataclasses import dataclass
from typing import Dict, List, Tuple, Optional

import numpy as np
import pandas as pd
from scipy.stats import ttest_1samp
from sklearn.model_selection import ParameterSampler
from sklearn.preprocessing import StandardScaler

import torch
from stable_baselines3 import PPO
from stable_baselines3.common.env_util import make_vec_env
from stable_baselines3.common.vec_env import VecNormalize
# Add once near imports
# ===== ADD: Monthly macro-step wrapper =====
import gymnasium as gym
import numpy as np

class MonthlyMacroWrapper(gym.Wrapper):
    """Each .step() advances a full month internally.
    The agent acts on the rebalance day; wrapper forwards zeros for the other days.
    """
    def __init__(self, env, rebalance_period: int):
        super().__init__(env)
        self.rebalance_period = int(rebalance_period)
        self._n = env.action_space.shape[0]

    def reset(self, **kwargs):
        return self.env.reset(**kwargs)

    def step(self, action):
        obs, reward, done, trunc, info = self.env.step(action)  # rebalance day
        for _ in range(self.rebalance_period - 1):
            if done:
                break
            zeros = np.zeros(self._n, dtype=np.float32)
            obs, r, done, trunc, info = self.env.step(zeros)
            reward += r  # monthly reward is non-zero on the last day
        return obs, reward, done, trunc, info

import numpy as np
import pandas as pd

def add_safe_history_features(
    df: pd.DataFrame,
    etf_list,
    *,
    ret_lags=(1,),                 # which past-day returns to include
    trailing_windows=(5, 21),      # trailing mean/vol windows (ending at t-1)
    add_price_lags=False,          # set True if you want price lags too
    price_lags=(1,),               # which price lags to include
    prefer_actual_return=True      # use Actual_Return_* if present; else derive from Price_*
) -> pd.DataFrame:
    """
    Adds features that are *only* functions of <= t-1 information:
      - RetLag_k_<T>: return from t-k
      - RetMean_W_<T>: mean of returns over [t-W, ..., t-1]
      - RetVol_W_<T>:  std of returns over [t-W, ..., t-1]
      - (optional) PriceLag_k_<T>, LogPriceLag_k_<T>
    Does NOT modify Actual_Return_*; env can still use it for rewards.
    """
    out = df.copy()
    for e in etf_list:
        # choose source for realized returns
        r = out.get(f"Actual_Return_{e}")
        if (r is None) or (not prefer_actual_return):
            p = out.get(f"Price_{e}")
            if p is None:
                continue
            r = p.pct_change()

        # simple lags
        for k in ret_lags:
            out[f"RetLag_{k}_{e}"] = r.shift(k)

        # trailing stats up to t-1
        r_past = r.shift(1)  # ensure the window ends at t-1
        for w in trailing_windows:
            out[f"RetMean_{w}_{e}"] = r_past.rolling(w, min_periods=max(2, w//2)).mean()
            out[f"RetVol_{w}_{e}"]  = r_past.rolling(w, min_periods=max(2, w//2)).std()

        # optional price lags (safe only if lag>=1)
        if add_price_lags:
            p = out.get(f"Price_{e}")
            if p is not None:
                for k in price_lags:
                    out[f"PriceLag_{k}_{e}"]    = p.shift(k)
                    out[f"LogPriceLag_{k}_{e}"] = np.log(p).shift(k)

    return out

# ===== ADD: VecNormalize in-memory cloning (optional speed-up) =====
from copy import deepcopy
from stable_baselines3.common.vec_env import VecNormalize

# ===== ADD: technical indicators exactly as Stage‑1 formulas =====
import ta

# Per‑asset column resolver: supports BOTH "Base_ETF" and "ETF_Base"


def copy_vecnorm_stats(src, dst, use_vecnormalize: bool):
    """Copy vecnormalize stats only when both envs are VecNormalize."""
    if not use_vecnormalize:
        return
    if not isinstance(src, VecNormalize) or not isinstance(dst, VecNormalize):
        return
    # In norm_obs=False, obs_rms may be None/unused. Guard everything.
    try:
        if hasattr(src, "obs_rms") and src.obs_rms is not None:
            dst.obs_rms = deepcopy(src.obs_rms)
        if hasattr(src, "ret_rms") and src.ret_rms is not None:
            dst.ret_rms = deepcopy(src.ret_rms)
        if hasattr(src, "clip_obs"):
            dst.clip_obs = src.clip_obs
        if hasattr(src, "clip_reward"):
            dst.clip_reward = src.clip_reward
        if hasattr(src, "gamma"):
            dst.gamma = src.gamma
        dst.training = False
    except AttributeError:
        # If dst is not VecNormalize after all, just skip copying
        return

import numpy as np
import pandas as pd

def _resolve_factor(df: pd.DataFrame, canonical: str, variants: list[str]) -> pd.Series | None:
    """Resolve a factor column by tolerant matching (case, separators)."""
    # exact first
    for v in variants:
        if v in df.columns:
            return df[v]
    # tolerant search
    norm = {c.lower().replace("-", "").replace("_", ""): c for c in df.columns}
    keys = [canonical] + variants
    for k in keys:
        k_ = k.lower().replace("-", "").replace("_", "")
        if k_ in norm:
            return df[norm[k_]]
    return None

def _load_prices_rename(prices_path: str, etf_list: list[str]) -> pd.DataFrame:
    """Load prices, rename ticker cols to Price_<TICKER>, return indexed by Date."""
    pr = pd.read_csv(prices_path, parse_dates=["Date"])
    pr["Date"] = pd.to_datetime(pr["Date"], utc=True).dt.tz_localize(None)
    # rename raw ticker cols -> Price_<TICKER>
    ren = {c: (f"Price_{c}" if c != "Date" and not str(c).startswith("Price_") else c) for c in pr.columns}
    pr = pr.rename(columns=ren).set_index("Date").sort_index()
    # keep only requested ETFs that exist
    keep = [f"Price_{e}" for e in etf_list if f"Price_{e}" in pr.columns]
    missing = [e for e in etf_list if f"Price_{e}" not in pr.columns]
    if missing:
        print(f"[WARN] Missing prices for: {missing}. Skipping those ETFs for per‑ETF indicators.")
    return pr[keep]

def _load_factors_canonical(factors_path: str) -> pd.DataFrame:
    """Load factors and return a frame with canonical columns when possible."""
    fa = pd.read_csv(factors_path, parse_dates=["Date"])
    fa["Date"] = pd.to_datetime(fa["Date"], utc=True).dt.tz_localize(None)
    fa = fa.set_index("Date").sort_index()

    # Try to resolve FF5 names
    rf   = _resolve_factor(fa, "RF",      ["RF","RiskFree","risk_free","R_f"])
    mktrf= _resolve_factor(fa, "Mkt-RF",  ["MKT-RF","Mkt_RF","MKT_RF","MKTminusRF","MktMinusRF"])
    mkt  = _resolve_factor(fa, "MKT",     ["MKT","Mkt","Market","MKTret","Mkt_ret"])
    smb  = _resolve_factor(fa, "SMB",     ["SMB","smb"])
    hml  = _resolve_factor(fa, "HML",     ["HML","hml"])
    rmw  = _resolve_factor(fa, "RMW",     ["RMW","rmw"])
    cma  = _resolve_factor(fa, "CMA",     ["CMA","cma"])

    out = pd.DataFrame(index=fa.index)
    if mktrf is not None:
        out["Mkt-RF"] = mktrf
    elif (mkt is not None) and (rf is not None):
        out["Mkt-RF"] = mkt - rf
    else:
        print("[WARN] Could not resolve Mkt‑RF; it will be missing.")

    if cma is not None: out["CMA"] = cma
    if smb is not None: out["SMB"] = smb
    if rmw is not None: out["RMW"] = rmw
    if hml is not None: out["HML"] = hml  # not used here, but harmless

    return out

def make_equal_weight_baseline_fn(etf_list):
    """
    Baseline for no-Stage1 benchmark: static equal-weight portfolio.
    """
    N = len(etf_list)
    ew = np.ones(N, dtype=float) / N

    def baseline_fn(date):
        # Ignores date; always returns equal-weight
        return ew

    return baseline_fn

def make_finrl_baseline_fn(df, etf_list):
    """
    FinRL-style baseline with seed automatically inherited from the global RNG.
    - No need to modify make_env_from_frame
    - The baseline will use the same seed as the current training iteration
    """
    n_assets = len(etf_list)

    # IMPORTANT:
    # Use the existing global numpy RNG so the baseline uses the same
    # random seed that your training loop already set.
    rng = np.random

    cache = {}

    def _softmax(x):
        x = np.asarray(x, float)
        x_shift = x - np.max(x)
        ex = np.exp(x_shift)
        denom = ex.sum()
        if denom <= 0 or not np.isfinite(denom):
            return np.ones_like(x) / len(x)
        return ex / denom

    def baseline_fn(date):
        if date not in cache:
            raw = rng.standard_normal(n_assets)
            w = _softmax(raw)
            cache[date] = w
        return cache[date]

    return baseline_fn

from typing import List, Callable, Dict
import numpy as np
import pandas as pd

# ===== ADD: Tilt baseline built from Stage-1 predicted returns =====
def make_dia_stage1_baseline_fn(
    df: pd.DataFrame,
    etf_list: List[str],
    beta: float = 0.10,
    vol_col_prefix: str = "Vol_20_",
    use_inv_vol: bool = True,
    ema_halflife_months: float = 3.0,
    floor: float = 0.02,
    cap: float = 0.22,
) -> Callable[[pd.Timestamp], np.ndarray]:
    """
    DIA universe static baseline using Stage-1 factors:

    Stage-1 factor set:
        ["ATR", "CMA_lag_2", "EMA_12", "HML_lag_2",
         "LagRet_1", "LagRet_2", "MACD", "Mkt-RF",
         "Mom_3", "SMB_lag_2"]

    - If Predicted_Return_* exists: score_mat = Predicted_Return_*.
    - Else build a composite from:
        * Per-asset indicators:
            LagRet_1, LagRet_2, Mom_3, MACD,
            EMA_12: use Price/EMA_12 - 1
            ATR:    penalty (higher ATR -> lower score)
        * Global factors (broadcast):
            CMA_lag_2, HML_lag_2, Mkt-RF, SMB_lag_2
      Each component is cross-sectionally z-scored per date, then averaged.
    - Then same inverse-vol scaling + linear tilt + floor/cap + EMA smoothing.
    """
    dates = df["Date"].to_list()
    T, N = len(dates), len(etf_list)

    # ---------- helpers ----------
    def _any_col(df_, etf, base):
        c1, c2 = f"{base}_{etf}", f"{etf}_{base}"
        if c1 in df_.columns:
            return df_[c1]
        if c2 in df_.columns:
            return df_[c2]
        return None

    def cs_zscore(M: np.ndarray) -> np.ndarray:
        mu = np.nanmean(M, axis=1, keepdims=True)
        sd = np.nanstd(M, axis=1, keepdims=True) + 1e-8
        Z  = (M - mu) / sd
        return np.nan_to_num(Z, nan=0.0, posinf=0.0, neginf=0.0)

    def mat_from_per_asset(base: str) -> np.ndarray:
        M = np.full((T, N), np.nan, dtype=float)
        for j, etf in enumerate(etf_list):
            s = _any_col(df, etf, base)
            if s is not None:
                M[:, j] = s.to_numpy(float)
        return M

    def mat_from_global(base: str) -> np.ndarray:
        if base not in df.columns:
            return np.full((T, N), np.nan, dtype=float)
        v = df[base].to_numpy(float)[:, None]
        return np.repeat(v, N, axis=1)

    # ---------- choose score matrix ----------
    pred_cols = [f"Predicted_Return_{e}" for e in etf_list]
    has_pred  = all(c in df.columns for c in pred_cols)

    if has_pred:
        score_mat = df[pred_cols].to_numpy(float)
    else:
        parts: List[np.ndarray] = []

        # DIA Stage-1 set:
        # ["ATR", "CMA_lag_2", "EMA_12", "HML_lag_2",
        #  "LagRet_1", "LagRet_2", "MACD", "Mkt-RF",
        #  "Mom_3", "SMB_lag_2"]

        # 1) Per-asset momentum/lag: LagRet_1, LagRet_2, Mom_3
        for base in ["LagRet_1", "LagRet_2", "Mom_3"]:
            M = mat_from_per_asset(base)
            if np.isfinite(M).any():
                parts.append(cs_zscore(M))

        # 2) MACD
        M_macd = mat_from_per_asset("MACD")
        if np.isfinite(M_macd).any():
            parts.append(cs_zscore(M_macd))

        # 3) EMA_12: Price / EMA_12 - 1
        M_ema = np.full((T, N), np.nan, dtype=float)
        for j, etf in enumerate(etf_list):
            p   = df.get(f"Price_{etf}")
            ema = _any_col(df, etf, "EMA_12")
            if p is not None and ema is not None:
                M_ema[:, j] = (p / ema - 1.0).to_numpy(float)
        if np.isfinite(M_ema).any():
            parts.append(cs_zscore(M_ema))

        # 4) ATR: penalty (higher ATR -> lower score)
        M_atr = mat_from_per_asset("ATR")
        if np.isfinite(M_atr).any():
            parts.append(-cs_zscore(M_atr))

        # 5) Global factors: CMA_lag_2, HML_lag_2, Mkt-RF, SMB_lag_2
        for base in ["CMA_lag_2", "HML_lag_2", "Mkt-RF", "SMB_lag_2"]:
            M = mat_from_global(base)
            if np.isfinite(M).any():
                parts.append(cs_zscore(M))

        if parts:
            score_mat = np.nanmean(np.stack(parts, axis=2), axis=2)
            score_mat = np.nan_to_num(score_mat, nan=0.0)
        else:
            score_mat = np.zeros((T, N), dtype=float)

    # ---------- inverse-vol scaling ----------
    vol_cols = [f"{vol_col_prefix}{e}" for e in etf_list]
    have_vol = all(c in df.columns for c in vol_cols)
    if use_inv_vol and have_vol:
        vol_mat = np.clip(df[vol_cols].to_numpy(float), 1e-6, None)
    else:
        vol_mat = np.ones_like(score_mat)

    # ---------- linear tilt + EMA + floor/cap ----------
    def _project_floor_cap(w, lo=floor, hi=cap):
        w = np.clip(w, lo, hi)
        s = w.sum()
        return w / s if s > 0 else np.ones_like(w) / len(w)

    lam = 0.0 if not ema_halflife_months or ema_halflife_months <= 0 \
        else 1 - 0.5 ** (1.0 / ema_halflife_months)

    baseline_cache: Dict[pd.Timestamp, np.ndarray] = {}
    prev_w = np.ones(N) / N

    for i, d in enumerate(dates):
        x = score_mat[i] / vol_mat[i]
        m, s = np.nanmean(x), np.nanstd(x)
        if not np.isfinite(s) or s < 1e-8:
            z = np.zeros_like(x)
        else:
            z = (x - m) / s
        z = np.nan_to_num(z, nan=0.0)

        # Linear tilt around equal-weight
        w_raw = np.ones(N) / N + beta * z

        # Long-only
        w_pos = np.clip(w_raw, 0.0, None)
        if w_pos.sum() <= 0:
            w0 = np.ones(N) / N
        else:
            w0 = w_pos / w_pos.sum()

        # Floor/cap
        w0 = _project_floor_cap(w0)

        # EMA smoothing
        if lam > 0:
            w = (1 - lam) * prev_w + lam * w0
            w = _project_floor_cap(w)
        else:
            w = w0

        baseline_cache[d] = w
        prev_w = w

    def baseline_fn(date):
        return baseline_cache.get(date, np.ones(N) / N)

    return baseline_fn
from typing import Callable, Dict, List
import numpy as np
import pandas as pd


from typing import Callable, Dict, List
import numpy as np
import pandas as pd


def make_dia_momentum_benchmark_fn(
    df: pd.DataFrame,
    stock_list: List[str],
    beta: float = 0.10,
    beta_scale: float = 1.0,                 # <-- ADDED

    # ---- match sector benchmark structure ----
    use_score_ewma: bool = True,
    score_ewma_halflife_days: float = 3.0,

    vol_col_prefix: str = "Vol_20_",
    use_inv_vol: bool = False,

    # optional weight EMA (default OFF)
    use_weight_ema: bool = False,
    ema_halflife_months: float = 3.0,
    trading_days_per_month: int = 21,

    # constraints (stocks usually tighter than ETFs)
    floor: float = 0.00,
    cap: float = 0.15,
) -> Callable[[pd.Timestamp], np.ndarray]:
    """
    DIA benchmark baseline (Stage-1-free): momentum-only tilt around equal-weight.

    Uses momentum_basic indicators per stock:
      - Mom_5, Mom_10, Mom_20 : positive contribution
      - Vol_20                : negative contribution

    Pattern mirrors your sector ETF baseline style:
      1) z-score each component cross-sectionally per day
      2) average components -> raw_score
      3) pandas daily cs z-score (mu/sd) each day
      4) optional EWMA on standardized signal
      5) optional inv-vol scaling using Vol_20_{ticker}
      6) final daily cs z-score -> linear tilt -> long-only -> floor/cap
      7) optional weight EMA (off by default)

    beta_eff = beta * beta_scale
    """
    if "Date" not in df.columns:
        raise KeyError("df must contain a 'Date' column.")
    dates = pd.to_datetime(df["Date"]).to_list()
    T, N = len(dates), len(stock_list)

    # ---------- helpers ----------
    def _any_col(df_: pd.DataFrame, tk: str, base: str):
        c1, c2 = f"{base}_{tk}", f"{tk}_{base}"
        if c1 in df_.columns:
            return df_[c1]
        if c2 in df_.columns:
            return df_[c2]
        return None

    def mat_from_cols(base: str) -> np.ndarray:
        M = np.full((T, N), np.nan, dtype=float)
        for j, tk in enumerate(stock_list):
            s = _any_col(df, tk, base)
            if s is not None:
                M[:, j] = s.to_numpy(float)
        return M

    def cs_zscore_np(M: np.ndarray) -> np.ndarray:
        mu = np.nanmean(M, axis=1, keepdims=True)
        sd = np.nanstd(M, axis=1, keepdims=True) + 1e-8
        Z = (M - mu) / sd
        return np.nan_to_num(Z, nan=0.0, posinf=0.0, neginf=0.0)

    def _project_floor_cap(w: np.ndarray, lo: float = floor, hi: float = cap) -> np.ndarray:
        w = np.clip(w, lo, hi)
        s = float(w.sum())
        return (w / s) if s > 0 else np.ones_like(w) / len(w)

    # ---------- build momentum-based raw score ----------
    parts: List[np.ndarray] = []

    for base in ["Mom_5", "Mom_10", "Mom_20"]:
        M = mat_from_cols(base)
        if np.isfinite(M).any():
            parts.append(cs_zscore_np(M))

    M_vol20 = mat_from_cols("Vol_20")
    if np.isfinite(M_vol20).any():
        parts.append(-cs_zscore_np(M_vol20))

    if parts:
        raw_score = np.nanmean(np.stack(parts, axis=2), axis=2)
        raw_score = np.nan_to_num(raw_score, nan=0.0)
    else:
        raw_score = np.zeros((T, N), dtype=float)

    # ---------- EXACT SAME PANDAS PATTERN AS YOUR SECTOR MODEL (standardize -> EWMA) ----------
    score_df = pd.DataFrame(raw_score)

    mu = score_df.mean(axis=1)
    sd = score_df.std(axis=1).replace(0, np.nan)
    z = score_df.sub(mu, axis=0).div(sd, axis=0).fillna(0.0)

    if use_score_ewma and (score_ewma_halflife_days is not None) and (score_ewma_halflife_days > 0):
        score_mat = (
            z.ewm(
                halflife=float(score_ewma_halflife_days),
                adjust=False,
                min_periods=1,
            )
            .mean()
            .to_numpy(float)
        )
    else:
        score_mat = z.to_numpy(float)

    # ---------- optional inverse-vol scaling ----------
    vol_cols = [f"{vol_col_prefix}{tk}" for tk in stock_list]
    have_vol = all(c in df.columns for c in vol_cols)
    if use_inv_vol and have_vol:
        vol_mat = np.clip(df[vol_cols].to_numpy(float), 1e-6, None)
    else:
        vol_mat = np.ones_like(score_mat)

    # ---------- optional EMA on weights ----------
    if use_weight_ema and (ema_halflife_months is not None) and (ema_halflife_months > 0):
        halflife_days = float(ema_halflife_months) * float(trading_days_per_month)
        lam = 1.0 - 0.5 ** (1.0 / halflife_days)
    else:
        lam = 0.0

    beta_eff = float(beta) * float(beta_scale)

    baseline_cache: Dict[pd.Timestamp, np.ndarray] = {}
    prev_w = np.ones(N, dtype=float) / N

    for i, d in enumerate(dates):
        x = score_mat[i] / vol_mat[i]

        m, s = float(np.nanmean(x)), float(np.nanstd(x))
        if (not np.isfinite(s)) or (s < 1e-8):
            zt = np.zeros_like(x)
        else:
            zt = (x - m) / s
        zt = np.nan_to_num(zt, nan=0.0, posinf=0.0, neginf=0.0)

        # linear tilt around equal-weight
        w_raw = np.ones(N) / N + beta_eff * zt

        # long-only normalize
        w_pos = np.clip(w_raw, 0.0, None)
        w0 = (w_pos / w_pos.sum()) if w_pos.sum() > 0 else np.ones(N) / N

        # floor/cap projection
        w0 = _project_floor_cap(w0)

        # optional weight EMA smoothing
        if lam > 0:
            w = (1.0 - lam) * prev_w + lam * w0
            w = _project_floor_cap(w)
        else:
            w = w0

        baseline_cache[d] = w
        prev_w = w

    def baseline_fn(date: pd.Timestamp) -> np.ndarray:
        return baseline_cache.get(pd.to_datetime(date), np.ones(N) / N)

    return baseline_fn


# 
# def make_signal_baseline_fn(
#     df: pd.DataFrame,
#     etf_list: List[str],
#     beta: float = 0.10,
#     vol_col_prefix: str = "Vol_20_",
#     use_inv_vol: bool = True,
#     ema_halflife_months: float = 3.0,
#     floor: float = 0.02,
#     cap: float = 0.22,
# ) -> callable:
#     """
#     Simplified baseline using SAME score inputs as v2:
#       - score_mat: Predicted_Return_* if present, else composite Stage-1 indicators
#       - optional inverse-vol scaling
#       - cross-sectional z-score
#       - linear tilt around equal-weight with fixed beta
#       - long-only + floor/cap + EMA smoothing
#     """
#     dates = df["Date"].to_list()
#     T, N = len(dates), len(etf_list)
# 
#     # ---------- choose score matrix (UNCHANGED from v2) ----------
#     pred_cols = [f"Predicted_Return_{e}" for e in etf_list]
#     has_pred  = all(c in df.columns for c in pred_cols)
# 
#     if has_pred:
#         score_mat = df[pred_cols].to_numpy(float)                        # [T, N]
#     else:
#         # === Fallback: parsimonious composite of standard technical signals ===
#         def _any_col(df_, etf, base):
#             c1, c2 = f"{base}_{etf}", f"{etf}_{base}"
#             if c1 in df_.columns: return df_[c1]
#             if c2 in df_.columns: return df_[c2]
#             return None
# 
#         def mat_from_cols(bases: list[str]):
#             M = np.full((T, N), np.nan, dtype=float)
#             for j, etf in enumerate(etf_list):
#                 for base in bases:
#                     s = _any_col(df, etf, base)
#                     if s is not None:
#                         M[:, j] = s.to_numpy(float)
#                         break
#             return M
# 
#         def cs_zscore(M):
#             mu = np.nanmean(M, axis=1, keepdims=True)
#             sd = np.nanstd(M,  axis=1, keepdims=True) + 1e-8
#             Z = (M - mu) / sd
#             return np.nan_to_num(Z, nan=0.0, posinf=0.0, neginf=0.0)
# 
#         parts = []
# 
#         # 1) Short-term momentum / past return (Mom_3 and LagRet_1)
#         M_mom3 = mat_from_cols(["Mom_3"])
#         if np.isfinite(M_mom3).any():
#             parts.append(cs_zscore(M_mom3))
# 
#         M_lag1 = mat_from_cols(["LagRet_1"])
#         if np.isfinite(M_lag1).any():
#             parts.append(cs_zscore(M_lag1))
# 
#         # 2) Trend: price vs short moving average (Price/SMA_5 - 1)
#         M_trend = np.full((T, N), np.nan, dtype=float)
#         for j, etf in enumerate(etf_list):
#             p   = df.get(f"Price_{etf}")
#             sma = _any_col(df, etf, "SMA_5")
#             if p is not None and sma is not None:
#                 M_trend[:, j] = (p / sma - 1.0).to_numpy(float)
#         if np.isfinite(M_trend).any():
#             parts.append(cs_zscore(M_trend))
# 
#         # 3) Overbought / oversold: RSI_7 (centered at 50)
#         M_rsi = mat_from_cols(["RSI_7"])
#         if np.isfinite(M_rsi).any():
#             parts.append(cs_zscore(M_rsi - 50.0))
# 
#         # 4) Volatility penalty: higher Vol_5 -> lower score
#         M_vol5 = mat_from_cols(["Vol_5"])
#         if np.isfinite(M_vol5).any():
#             parts.append(-cs_zscore(M_vol5))
# 
#         # Combine signals
#         if parts:
#             score_mat = np.nanmean(np.stack(parts, axis=2), axis=2)      # [T, N]
#             score_mat = np.nan_to_num(score_mat, nan=0.0)
#         else:
#             score_mat = np.zeros((T, N), dtype=float)
# 
#     # ---------- inverse-vol scaling (optional, unchanged) ----------
#     vol_cols = [f"{vol_col_prefix}{e}" for e in etf_list]
#     have_vol = all(c in df.columns for c in vol_cols)
#     if use_inv_vol and have_vol:
#         vol_mat = np.clip(df[vol_cols].to_numpy(float), 1e-6, None)     # [T, N]
#     else:
#         vol_mat = np.ones_like(score_mat)
# 
#     # ---------- linear tilt + EMA + floor/cap ----------
#     def _project_floor_cap(w, lo=floor, hi=cap):
#         w = np.clip(w, lo, hi)
#         s = w.sum()
#         return w / s if s > 0 else np.ones_like(w)/len(w)
# 
#     lam = 0.0 if not ema_halflife_months or ema_halflife_months <= 0 \
#         else 1 - 0.5 ** (1.0 / ema_halflife_months)
# 
#     baseline_cache = {}
#     prev_w = np.ones(N) / N
# 
#     for i, d in enumerate(dates):
#         x = score_mat[i] / vol_mat[i]
#         m, s = np.nanmean(x), np.nanstd(x)
#         if not np.isfinite(s) or s < 1e-8:
#             z = np.zeros_like(x)
#         else:
#             z = (x - m) / s
#         z = np.nan_to_num(z, nan=0.0)
# 
#         # linear tilt around equal-weight
#         w_raw = np.ones(N) / N + beta * z
# 
#         # project to long-only
#         w_pos = np.clip(w_raw, 0.0, None)
#         if w_pos.sum() <= 0:
#             w0 = np.ones(N) / N
#         else:
#             w0 = w_pos / w_pos.sum()
# 
#         # floor/cap
#         w0 = _project_floor_cap(w0)
# 
#         # EMA smoothing
#         if lam > 0:
#             w = (1 - lam) * prev_w + lam * w0
#             w = _project_floor_cap(w)
#         else:
#             w = w0
# 
#         baseline_cache[d] = w
#         prev_w = w
# 
#     def baseline_fn(date):
#         return baseline_cache.get(date, np.ones(N)/N)
# 
#     return baseline_fn



# -------------------------
# Config
# -------------------------
@dataclass
class TrainingConfig:
    # Windows
    train_window_days: int = 252 * 7
    validation_window_days: int = 252
    prediction_window_days: int = 252

    # Observation & action cadence
    lookback_period: int = 21
    rebalance_period: int = 21
    macro_step: bool = True        # << monthly macro-steps

    # PPO budgets (monthly steps)
    n_iter_tuning: int = 8
    tuning_timesteps: int = 1_500
    incremental_timesteps: int = 2_000
    max_timesteps: int = 12_000
    patience: int = 2

    # Network
    policy_arch: Tuple[int, int] = (64, 64)
    n_epochs: int = 5
    device: str = "auto"           # "cuda" | "cpu" | "auto"

    # Iterations/seeds
    num_iterations: int = 30
    base_seed: int = 42

    # Env economics (slightly looser for more variation)
    default_risk_coeff: float = 0.05
    desired_long: float = 1.0
    desired_short: float = 0.0
    weight_bounds: Tuple[float, float] = (0.0  , 1.0)
    lambda_hhi: float = 0.04        # ↓ from 0.1
    lambda_turnover: float = 0.003  # ↓ from 0.005
    transaction_cost_rate: float = 0.0005
    
    # lambda_turnover: float = 0.0  # ↓ from 0.005
    # lambda_hhi: float = 0.0        # ↓ from 0.1
    # transaction_cost_rate: float = 0.0
    
    alpha: float = 0.05
    cvar_warmup_months: int = 12
    action_temperature: float = 0.4 # ↓ from 0.6: clearer tilts
    exclude_price_levels: bool = True
    allow_actual_returns: bool = False 

    # VecNormalize
    use_vecnormalize: bool = True
    vecnorm_clip_obs: float = 10.0

    # Model reuse
    warm_start_across_windows: bool = True

    use_robust_scaler_for_shap: bool = True
    robust_q_low: int = 5
    robust_q_high: int = 95
    shap_feature_gain: float = 1.3   # 1.0 = off; try 1.1–1.5
    shap_dropout_p: float = 0.10      # feature dropout on SHAP (train only)
    
    use_compact_shap: bool = False 
    shap_align_coef: float = 0.0 # (0.01–0.05 works).
    
    include_technical_features: bool = True      # <— master switch
    baseline_use_inv_vol: bool = False            # use 1/vol in baseline if vol cols exist
    baseline_vol_prefix: str = "Vol_20_"  
    tech_set : str = "momentum_basic" # default momentum_basic | stage1_indicators


# -------------------------
# Utilities
# -------------------------

from sklearn.preprocessing import StandardScaler, RobustScaler

def get_shap_cols(cols):
    return [c for c in cols if c.startswith(("SHAP_DIR_", "SHAP_CONF_", "SHAP_DIR_RESID_", "SHAP_PRIOR_"))]

def apply_shap_conf_gate_inplace(df, etf_list, tau=0.0):
    # Keep SHAP direction only when confidence is positive & above tau
    for etf in etf_list:
        d = f"SHAP_DIR_{etf}_CSZ"
        c = f"SHAP_CONF_{etf}_CSZ"
        if d in df.columns and c in df.columns:
            m = (df[c] <= tau)  # gate at tau (0 is a good start)
            df.loc[m, d] = 0.0
            # Optional “prior” channel = gated, signed confidence
            df[f"SHAP_PRIOR_{etf}"] = df[d] * df[c].clip(lower=0.0)

def shap_dropout_inplace(df, shap_cols, p, rng):
    if p <= 0 or not shap_cols:
        return
    mask = rng.random(size=(len(df), len(shap_cols))) < p
    # drop to zero (acts like “no SHAP info” for those samples)
    df.loc[:, shap_cols] = df[shap_cols].mask(mask, other=0.0)
    
def set_global_seed(seed: int) -> None:
    np.random.seed(seed)
    random.seed(seed)
    torch.manual_seed(seed)

# ADD near imports (once)
import ta

# Helpers to tolerate both "Base_ETF" and "ETF_Base" naming
def _get_per_asset_series(df: pd.DataFrame, etf: str, base: str):
    c1 = f"{base}_{etf}"     # our Stage‑2 canonical
    c2 = f"{etf}_{base}"     # possible Stage‑1 export naming
    if c1 in df.columns: return df[c1]
    if c2 in df.columns: return df[c2]
    return None

def _put_per_asset_col(cols_dict: dict, etf: str, base: str, s: pd.Series):
    # Always publish as "Base_ETF" to keep Stage‑2 consistent
    cols_dict[f"{base}_{etf}"] = s


def add_stable_features(df: pd.DataFrame,
                        etf_list: list[str],
                        enabled: bool = True,
                        tech_set: str = "stage1_indicators",
                        *,
                        prices_path: str = "/mnt/data/stock_prices_10ETFs.csv",
                        factors_path: str = "/mnt/data/aligned_factors.csv",
                        stage1_only: bool = True,
                        indicators: list[str] | None = None,          # for stage1_indicators
                        momentum_indicators: list[str] | None = None  # for momentum_basic
                        ) -> pd.DataFrame:
    """
    Build Stage‑2 inputs while PRESERVING all existing columns in df.
    Will compute per‑asset indicators with the SAME formulas used in Stage‑1 (ta.*),
    and will re‑use any precomputed <ETF>_<FEATURE> columns if present.

    tech_set modes (case/space insensitive):
      - "stage1_indicators" (aliases: "stage1", "dia_stage1")
      - "momentum_basic"    (alias: "momentum")
    """
    if not enabled:
        return df.copy()

    if "Date" not in df.columns:
        raise ValueError("Input df must contain a 'Date' column.")

    # Robust mode canonicalization
    mode = str(tech_set).strip().lower()
    if mode in {"stage1", "dia_stage1"}:
        mode = "stage1_indicators"
    if mode == "momentum":
        mode = "momentum_basic"

    out = df.copy()
    out["Date"] = pd.to_datetime(out["Date"], utc=True).dt.tz_localize(None)
    idx = out["Date"].sort_values().unique()

    # Price panel (Close) as Price_<TICKER>
    P = _load_prices_rename(prices_path, etf_list).reindex(idx)  # -> Price_<ETF>

    # Canonicalizers for indicator names
    def _canon_stage1(x: str) -> str:
        s = str(x).strip()
        k = s.lower().replace("-", "_")
        if k in {"mkt_rf", "mktrf"}:            return "Mkt-RF"
        if k in {"ema12", "ema_12"}:            return "EMA_12"
        if k in {"macd"}:                       return "MACD"
        if k in {"atr"}:                        return "ATR"
        if k in {"lagret1","lag_ret_1","lag_1"}:  return "LagRet_1"
        if k in {"lagret2","lag_ret_2","lag_2"}:  return "LagRet_2"
        if k in {"mom3","momentum3","momentum_3"}: return "Mom_3"
        if k in {"cma_lag2","cma_lag_2"}:       return "CMA_lag_2"
        if k in {"hml_lag2","hml_lag_2"}:       return "HML_lag_2"
        if k.startswith("smb") and "2" in k:    return "SMB_lag_2"
        if k in {"sma5","sma_5"}:               return "SMA_5"
        if k in {"rsi7","rsi_7"}:               return "RSI_7"
        if k in {"vol5","std5","stdev5"}:       return "Vol_5"
        return s

    def _canon_mom(x: str) -> str:
        s = str(x).strip().lower().replace("-", "_")
        if s in {"vol20","vol_20"}: return "Vol_20"
        if s in {"mom5","mom_5"}:   return "Mom_5"
        if s in {"mom10","mom_10"}: return "Mom_10"
        if s in {"mom20","mom_20"}: return "Mom_20"
        return x

    # =========================================================
    # STAGE‑1 PACK (DIA stocks)  ——  EXACT Stage‑1 formulas
    # =========================================================
    if mode == "stage1_indicators":
        # Your DIA feature set (add SMA/RSI/Vol_5 if you also want them around)
        if indicators is None:
            indicators = [
                "ATR", "CMA_lag_2", "EMA_12", "HML_lag_2",
                "LagRet_1", "LagRet_2", "MACD", "Mkt-RF",
                "Mom_3", "SMB_lag_2"
                # optional extras if desired downstream:
                # "SMA_5", "RSI_7", "Vol_5"
            ]
        sel = { _canon_stage1(x) for x in indicators }

        per_etf_allowed = {"EMA_12", "MACD", "ATR", "LagRet_1", "LagRet_2", "Mom_3", "SMA_5", "RSI_7", "Vol_5"}
        global_allowed  = {"Mkt-RF", "CMA_lag_2", "HML_lag_2", "SMB_lag_2"}

        # ---------- Global factors (apply lags here) ----------
        req_global = []
        if any(g in sel for g in global_allowed):
            F = _load_factors_canonical(factors_path).reindex(idx)
            feat_global = pd.DataFrame(index=idx)
            if "Mkt-RF" in sel and "Mkt-RF" in F.columns:
                feat_global["Mkt-RF"] = F["Mkt-RF"]; req_global.append("Mkt-RF")
            if "CMA_lag_2" in sel and "CMA" in F.columns:
                feat_global["CMA_lag_2"] = F["CMA"].shift(2); req_global.append("CMA_lag_2")
            if "HML_lag_2" in sel and "HML" in F.columns:
                feat_global["HML_lag_2"] = F["HML"].shift(2); req_global.append("HML_lag_2")
            if "SMB_lag_2" in sel and "SMB" in F.columns:
                feat_global["SMB_lag_2"] = F["SMB"].shift(2); req_global.append("SMB_lag_2")
            if len(feat_global.columns) > 0:
                out = out.merge(feat_global.reset_index().rename(columns={"index":"Date"}), on="Date", how="left")

        # ---------- Per‑asset indicators with ta (reuse if present) ----------
        req_per_etf, frames = [], []
        if any(p in sel for p in per_etf_allowed):
            for e in etf_list:
                pe = f"Price_{e}"
                if pe not in P.columns:
                    continue

                close = P[pe]
                ret   = close.pct_change()
                cols  = {}

                # EMA_12
                if "EMA_12" in sel:
                    s = _get_per_asset_series(out, e, "EMA_12")
                    if s is None:
                        s = ta.trend.ema_indicator(close, window=12)
                    _put_per_asset_col(cols, e, "EMA_12", s); req_per_etf.append(f"EMA_12_{e}")

                # MACD (12/26/9 diff)
                if "MACD" in sel:
                    s = _get_per_asset_series(out, e, "MACD")
                    if s is None:
                        s = ta.trend.macd_diff(close)
                    _put_per_asset_col(cols, e, "MACD", s); req_per_etf.append(f"MACD_{e}")

                # ATR proxy from Stage‑1 (high=1.01*close, low=0.99*close, window=10)
                if "ATR" in sel:
                    s = _get_per_asset_series(out, e, "ATR")
                    if s is None:
                        s = ta.volatility.average_true_range(
                            high=close * 1.01, low=close * 0.99, close=close, window=10
                        )
                    _put_per_asset_col(cols, e, "ATR", s); req_per_etf.append(f"ATR_{e}")

                # Lags & momentum
                if "LagRet_1" in sel:
                    s = _get_per_asset_series(out, e, "LagRet_1")
                    if s is None: s = ret.shift(1)
                    _put_per_asset_col(cols, e, "LagRet_1", s); req_per_etf.append(f"LagRet_1_{e}")

                if "LagRet_2" in sel:
                    s = _get_per_asset_series(out, e, "LagRet_2")
                    if s is None: s = ret.shift(2)
                    _put_per_asset_col(cols, e, "LagRet_2", s); req_per_etf.append(f"LagRet_2_{e}")

                if "Mom_3" in sel:
                    s = _get_per_asset_series(out, e, "Mom_3")
                    if s is None: s = ret.rolling(window=3).mean()
                    _put_per_asset_col(cols, e, "Mom_3", s); req_per_etf.append(f"Mom_3_{e}")

                # Optional extras (only if requested in indicators)
                if "SMA_5" in sel:
                    s = _get_per_asset_series(out, e, "SMA_5")
                    if s is None: s = ta.trend.sma_indicator(close, window=5)
                    _put_per_asset_col(cols, e, "SMA_5", s); req_per_etf.append(f"SMA_5_{e}")

                if "RSI_7" in sel:
                    s = _get_per_asset_series(out, e, "RSI_7")
                    if s is None: s = ta.momentum.rsi(close, window=7)
                    _put_per_asset_col(cols, e, "RSI_7", s); req_per_etf.append(f"RSI_7_{e}")

                if "Vol_5" in sel:
                    s = _get_per_asset_series(out, e, "Vol_5")
                    if s is None: s = ret.rolling(window=5).std()
                    _put_per_asset_col(cols, e, "Vol_5", s); req_per_etf.append(f"Vol_5_{e}")

                if cols:
                    frames.append(pd.DataFrame(cols, index=P.index))

            if frames:
                feat_etf = pd.concat(frames, axis=1)
                out = out.merge(feat_etf.reset_index().rename(columns={"index":"Date"}), on="Date", how="left")

        # Warm‑up drop only for newly created columns (don’t touch original df cols)
        req = req_global + req_per_etf
        if req:
            out = out.dropna(subset=[c for c in req if c in out.columns])

        return out

    # =========================================================
    # MOMENTUM PACK (optional, unchanged)
    # =========================================================
    if mode == "momentum_basic":
        if momentum_indicators is None:
            momentum_indicators = ["Vol_20", "Mom_5", "Mom_10", "Mom_20"]
        sel_m = { _canon_mom(x) for x in momentum_indicators }

        frames, req = [], []
        for e in etf_list:
            pe = f"Price_{e}"
            if pe not in P.columns:
                continue
            p = P[pe]; r = p.pct_change()
            cols = {}
            if "Vol_20" in sel_m: cols[f"Vol_20_{e}"] = r.rolling(20).std();  req.append(f"Vol_20_{e}")
            if "Mom_5"  in sel_m: cols[f"Mom_5_{e}"]  = r.rolling(5).mean();  req.append(f"Mom_5_{e}")
            if "Mom_10" in sel_m: cols[f"Mom_10_{e}"] = r.rolling(10).mean(); req.append(f"Mom_10_{e}")
            if "Mom_20" in sel_m: cols[f"Mom_20_{e}"] = r.rolling(20).mean(); req.append(f"Mom_20_{e}")
            if cols:
                frames.append(pd.DataFrame(cols, index=P.index))

        if frames:
            feat_etf = pd.concat(frames, axis=1)
            out = out.merge(feat_etf.reset_index().rename(columns={"index":"Date"}), on="Date", how="left")

        if req:
            out = out.dropna(subset=[c for c in req if c in out.columns])

        return out

    # Unknown mode
    raise ValueError(
        f"Unknown tech_set='{tech_set}'. Use 'stage1_indicators'/'stage1'/'dia_stage1' or 'momentum_basic'."
    )

import re
import numpy as np
import pandas as pd

# -------- SHAP processing helpers --------
def _find_pred_col_for_etf(cols, etf):
    # tries to find the Stage-1 predicted return column for a given etf
    # works with names like 'Predicted_Return_XLK', 'pred_xlk', etc.
    pat = re.compile(fr'(pred|forecast).*{etf}', flags=re.I)
    for c in cols:
        if pat.search(c):
            return c
    # common explicit naming
    cand = f"Predicted_Return_{etf}"
    return cand if cand in cols else None

def _shap_cols_for_etf(cols, etf):
    # grabs any SHAP column that contains the ETF token (either _XLK or XLK_)
    # so it works with SHAP_MKT_XLK or SHAP_XLK_MKT
    pat = re.compile(fr'(^|_)SHAP.*({etf})(_|$)', flags=re.I)
    return [c for c in cols if pat.search(c)]

def _rolling_residual(x: pd.Series, y: pd.Series, window=126):
    # residual of x after rolling OLS on y
    cov = y.rolling(window, min_periods=max(25, window//5)).cov(x)
    var = y.rolling(window, min_periods=max(25, window//5)).var()
    beta = cov / (var.replace(0, np.nan))
    mu_x = x.rolling(window, min_periods=max(25, window//5)).mean()
    mu_y = y.rolling(window, min_periods=max(25, window//5)).mean()
    alpha = mu_x - beta * mu_y
    resid = x - (alpha + beta * y)
    return resid.fillna(0.0)

def _cs_zscore(df: pd.DataFrame, cols):
    M = df[cols].values
    mean = np.nanmean(M, axis=1, keepdims=True)
    std  = np.nanstd(M,  axis=1, keepdims=True)
    Z = (M - mean) / (std + 1e-8)
    out = pd.DataFrame(Z, index=df.index, columns=[c + "_CSZ" for c in cols])
    return out

import re
import numpy as np
import pandas as pd

def build_compact_shap_features(df: pd.DataFrame,
                                etf_list,
                                ewm_span: int = 5,
                                orth_window: int = 126,
                                preserve_raw_if_no_etf: bool = True) -> pd.DataFrame:
    """
    Compact SHAP builder:
      - If per‑ETF SHAP columns exist, build SHAP_DIR_*_CSZ / SHAP_CONF_*_CSZ / SHAP_DIR_RESID_*.
      - If only aggregated SHAP exists (e.g., Avg_SHAP_*, Std_SHAP_*), leave SHAP as-is.
    """
    out = df.copy()
    cols = out.columns.tolist()

    # All SHAP‑like columns present
    shap_like = [c for c in cols if "shap" in c.lower()]
    if not shap_like:
        return out  # nothing to do

    # Helper: does a column mention a given ETF token?
    def _mentions_etf(c: str, etf: str) -> bool:
        # allow patterns like ..._XLK, XLK_, ...XLK..., etc.
        return bool(re.search(fr'(^|[_\-]){etf}([_\-]|$)', c, flags=re.I))

    # Collect per‑ETF SHAP columns
    per_etf_map = {etf: [c for c in shap_like if _mentions_etf(c, etf)] for etf in etf_list}
    has_any_per_etf = any(len(v) > 0 for v in per_etf_map.values())

    if not has_any_per_etf:
        # Only aggregated SHAP (Avg_SHAP_*, Std_SHAP_*, *_REL) detected.
        # Do NOT rename/drop; just keep raw SHAP.
        # (Optional) Could compute global SHAP_DIR/CONF—but not used for ETF alignment.
        return out

    # === Per‑ETF compaction path (unchanged logic, minimal) ===
    # Temporarily prefix raw SHAP so we don't collide
    rename_map = {c: f"RAW_{c}" for etf, lst in per_etf_map.items() for c in lst}
    out.rename(columns=rename_map, inplace=True)

    def _cs_zscore(df2: pd.DataFrame, columns):
        M = df2[columns].values
        mu = np.nanmean(M, axis=1, keepdims=True)
        sd = np.nanstd(M, axis=1, keepdims=True)
        Z = (M - mu) / (sd + 1e-8)
        return pd.DataFrame(Z, index=df2.index, columns=[c + "_CSZ" for c in columns])

    def _find_pred_col_for_etf(all_cols, etf):
        for c in all_cols:
            if c.lower().startswith("predicted_return_") and c.lower().endswith(etf.lower()):
                return c
        return None

    def _rolling_residual(x: pd.Series, y: pd.Series, window=126):
        cov = y.rolling(window, min_periods=max(25, window//5)).cov(x)
        var = y.rolling(window, min_periods=max(25, window//5)).var()
        beta = cov / (var.replace(0, np.nan))
        mu_x = x.rolling(window, min_periods=max(25, window//5)).mean()
        mu_y = y.rolling(window, min_periods=max(25, window//5)).mean()
        alpha = mu_x - beta * mu_y
        resid = x - (alpha + beta * mu_y)
        return resid.fillna(0.0)

    dir_cols, conf_cols = [], []
    for etf in etf_list:
        raw_cols = per_etf_map[etf]
        if not raw_cols:
            continue
        signed_sum = out[[f"RAW_{c}" for c in raw_cols]].sum(axis=1)
        abs_sum    = out[[f"RAW_{c}" for c in raw_cols]].abs().sum(axis=1)
        direction  = (signed_sum / (abs_sum.replace(0, np.nan))).fillna(0.0)
        conf_raw   = abs_sum
        conf_norm  = conf_raw / (conf_raw.rolling(252, min_periods=60).median().replace(0, np.nan))
        direction_sm = direction.ewm(span=ewm_span, adjust=False).mean()
        conf_sm      = conf_norm.ewm(span=ewm_span, adjust=False).mean()
        out[f"SHAP_DIR_{etf}"]  = direction_sm
        out[f"SHAP_CONF_{etf}"] = conf_sm
        dir_cols.append(f"SHAP_DIR_{etf}")
        conf_cols.append(f"SHAP_CONF_{etf}")

    add_frames = []
    if dir_cols:  add_frames.append(_cs_zscore(out, dir_cols))
    if conf_cols: add_frames.append(_cs_zscore(out, conf_cols))
    if add_frames:
        out = pd.concat([out] + add_frames, axis=1)

    # Orthogonalize direction vs predicted return
    for etf in etf_list:
        dir_csz = f"SHAP_DIR_{etf}_CSZ"
        pred    = _find_pred_col_for_etf(out.columns, etf)
        if dir_csz in out.columns and pred in out.columns:
            out[f"SHAP_DIR_RESID_{etf}"] = _rolling_residual(out[dir_csz], out[pred], window=orth_window)

    # Drop temporary RAW_ copies
    out.drop(columns=[c for c in out.columns if c.startswith("RAW_")], inplace=True, errors="ignore")
    return out
import re
import pandas as pd

def filter_features(
    df: pd.DataFrame,
    *,
    include_predicted_returns: bool = True,
    include_volatility: bool = True,
    include_stage1_indicators: bool = False,
    include_momentum_indicators: bool = False,
    include_shap_metrics: bool = True,
    shap_variant: str = "auto",
    include_crosssec_stats: bool = True,
    include_rank_predret: bool = True,
    include_safe_history_returns: bool = True,   # NEW
    include_safe_price_lags: bool = False,       # NEW
    drop_others: bool = False
) -> pd.DataFrame:
    cols = list(df.columns)
    keep = set()

    # Always keep Date + realized returns (for reward calc in env; will be stripped from obs by wrapper)
    keep.add("Date")
    keep.update([c for c in cols if c.startswith("Actual_Return_")])

    if include_predicted_returns:
        keep.update([c for c in cols if c.startswith("Predicted_Return_")])

    if include_volatility:
        keep.update([c for c in cols if c.startswith("Volatility_")])
        if "CrossSec_Mean_Volatility" in cols:
            keep.add("CrossSec_Mean_Volatility")

    if include_crosssec_stats:
        keep.update([c for c in cols if c.startswith("CrossSec_")])

    if include_rank_predret:
        keep.update([c for c in cols if c.startswith("Rank_PredRet_")])

    # SHAP
    compact_shap_cols = [
        c for c in cols
        if c.startswith("SHAP_DIR_")
        or c.startswith("SHAP_CONF_")
        or c.startswith("SHAP_DIR_RESID_")
        or c.endswith("_CSZ")
    ]
    raw_shap_cols = [c for c in cols if ("SHAP" in c and c not in compact_shap_cols)]
    if include_shap_metrics:
        sv = shap_variant.lower()
        if sv == "raw":   keep.update(raw_shap_cols)
        elif sv == "both": keep.update(raw_shap_cols + compact_shap_cols)
        elif sv == "compact": keep.update(compact_shap_cols)
        else: keep.update(compact_shap_cols if compact_shap_cols else raw_shap_cols)

    # Stage‑1 (DIA) — leave False if you want them out
    if include_stage1_indicators:
        for p in ("EMA_12_", "MACD_", "ATR_", "LagRet_1_", "LagRet_2_", "Mom_3_", "SMA_5_", "RSI_7_", "Vol_5_"):
            keep.update([c for c in cols if c.startswith(p)])
        for g in ("Mkt-RF", "CMA_lag_2", "HML_lag_2", "SMB_lag_2"):
            if g in cols: keep.add(g)

    # NEW: Safe history families
    if include_safe_history_returns:
        keep.update([c for c in cols if c.startswith(("RetLag_", "RetMean_", "RetVol_"))])

    if include_safe_price_lags:
        keep.update([c for c in cols if c.startswith(("PriceLag_", "LogPriceLag_"))])

    # ---------------- Produce final frame ----------------
    if drop_others:
        ordered_cols = [c for c in cols if c in keep]
        return df.loc[:, ordered_cols].copy()

    # else: only drop explicitly disabled families
    drop_cols = set()
    if not include_predicted_returns:
        drop_cols.update([c for c in cols if c.startswith("Predicted_Return_")])
    if not include_volatility:
        drop_cols.update([c for c in cols if c.startswith("Volatility_")])
        if "CrossSec_Mean_Volatility" in cols: drop_cols.add("CrossSec_Mean_Volatility")
    if not include_crosssec_stats:
        drop_cols.update([c for c in cols if c.startswith("CrossSec_")])
    if not include_rank_predret:
        drop_cols.update([c for c in cols if c.startswith("Rank_PredRet_")])
    if not include_shap_metrics:
        drop_cols.update([c for c in cols if "SHAP" in c])
    if not include_stage1_indicators:
        for p in ("EMA_12_", "MACD_", "ATR_", "LagRet_1_", "LagRet_2_", "Mom_3_", "SMA_5_", "RSI_7_", "Vol_5_"):
            drop_cols.update([c for c in cols if c.startswith(p)])
        for s in ("_EMA_12","_MACD","_ATR","_LagRet_1","_LagRet_2","_Mom_3","_SMA_5","_RSI_7","_Vol_5"):
            drop_cols.update([c for c in cols if c.endswith(s)])
        for g in ("Mkt-RF", "CMA_lag_2", "HML_lag_2", "SMB_lag_2"):
            if g in cols: drop_cols.add(g)
    if not include_safe_history_returns:
        drop_cols.update([c for c in cols if c.startswith(("RetLag_", "RetMean_", "RetVol_"))])
    if not include_safe_price_lags:
        drop_cols.update([c for c in cols if c.startswith(("PriceLag_", "LogPriceLag_"))])

    drop_cols = [c for c in drop_cols if c not in keep]
    return df.drop(columns=drop_cols, errors="ignore").copy()



def infer_etf_list(df: pd.DataFrame) -> List[str]:
    """Infer ETF universe from Actual_Return_* columns."""
    etfs = sorted([c.replace("Actual_Return_", "") for c in df.columns if c.startswith("Actual_Return_")])
    if not etfs:
        raise ValueError("No Actual_Return_* columns found to infer ETF list.")
    return etfs

# -------------------------
# Env builders (train/val/pred)
# -------------------------
def make_env_from_frame(df: pd.DataFrame,
                        etf_list: List[str],
                        cfg: TrainingConfig,
                        risk_coeff: float,
                        lambda_turnover: float,
                        lambda_hhi: float,
                        use_baseline: bool,
                        vecnorm_load_path: Optional[str] = None,
                        training: bool = True):
    """Build a (VecNormalized) env from a DataFrame slice."""
    def _ctor():
        if use_baseline:
        # Drop Stage-1 stuff so the baseline uses the composite technical signals
        
            # baseline_fn = make_finrl_baseline_fn(df, etf_list)
        
            df_baseline = df.drop(
                columns=[c for c in df.columns
                         if c.startswith("Predicted_Return_") or c.startswith("SHAP_")],
                errors="ignore",
            )

            # baseline_fn = make_signal_baseline_fn(
            #     df=df_baseline,
            #     etf_list=etf_list,
            #     beta=0.10,              # or your chosen value
            #     vol_col_prefix=cfg.baseline_vol_prefix,
            #     use_inv_vol=cfg.baseline_use_inv_vol,
            #     ema_halflife_months=3.0,
            #     floor=0.01,
            #     cap=1/3,
            # )
        
            # baseline_fn = make_dia_stage1_baseline_fn(
            #     df=df_baseline,
            #     etf_list=etf_list,
            #     beta=0.10,
            #     vol_col_prefix=cfg.baseline_vol_prefix,
            #     use_inv_vol=cfg.baseline_use_inv_vol,
            #     ema_halflife_months=3.0,
            #     floor=0.01,
            #     cap=1/3,
            # )
            baseline_fn = make_dia_momentum_benchmark_fn(
            df=df_baseline,
            stock_list=etf_list,
        
            # tilt strength
            beta=0.10,
            beta_scale=2.0,
            # signal smoothing (preferred)
            use_score_ewma=True,
            score_ewma_halflife_days=3.0,
        
            # inverse-vol (optional; often OFF for small-N ETFs)
            vol_col_prefix=cfg.baseline_vol_prefix,
            use_inv_vol=False,
        
            # avoid double smoothing
            use_weight_ema=False,
        
            # constraints
            floor=0.02,
            cap=1/3,
        )
        else:
            baseline_fn = None

        env = PortfolioEnv(
            df, etf_list,
            reward_type="mean_cvar",
            risk_coefficient=risk_coeff,
            rebalance_period=cfg.rebalance_period,
            lookback_period=cfg.lookback_period,
            weight_bounds=cfg.weight_bounds,
            desired_long=cfg.desired_long,
            desired_short=cfg.desired_short,
            use_baseline=use_baseline,
            baseline_fn=baseline_fn,
            transaction_cost_rate=cfg.transaction_cost_rate,
            lambda_turnover=lambda_turnover,
            lambda_hhi=lambda_hhi,
            alpha=cfg.alpha,
            cvar_warmup_months=cfg.cvar_warmup_months,
            exclude_price_levels=cfg.exclude_price_levels,
            allow_actual_returns_in_obs=cfg.allow_actual_returns,  
            action_temperature=cfg.action_temperature,
            shap_align_coef=(cfg.shap_align_coef if cfg.use_compact_shap else 0.0),  
        )
        if cfg.macro_step:
            env = MonthlyMacroWrapper(env, cfg.rebalance_period)
        return env

    venv = make_vec_env(_ctor, n_envs=1)

    if cfg.use_vecnormalize:
        if vecnorm_load_path is None:
            venv = VecNormalize(
                venv,
                norm_obs=False,      # IMPORTANT: keep SHAP scaling intact
                norm_reward=True,
                clip_obs=cfg.vecnorm_clip_obs
            )
        else:
            venv = VecNormalize.load(vecnorm_load_path, venv)
        venv.training = training


    return venv


def save_vecnorm_if_needed(venv, path: str, cfg: TrainingConfig):
    if cfg.use_vecnormalize:
        venv.save(path)

# -------------------------
# Tuning (smarter ranges)
# -------------------------
def sample_valid_params(cfg: TrainingConfig, random_seed: int, n: int):
    # Monthly steps: smaller, faster
    space = {
        "learning_rate": [3e-4, 2e-4, 1e-4],
        "n_steps": [64, 96, 128],
        "batch_size": [32, 64],
        "gamma": [0.96, 0.985, 0.995],
        "gae_lambda": [0.90, 0.95, 0.98],
        "ent_coef": [0.002, 0.005, 0.01],
        "clip_range": [0.1, 0.2],
        "vf_coef": [0.5, 0.7, 1.0],
        "max_grad_norm": [0.3, 0.5],
        # env economics
        # "risk_coefficient": [0.01, 0.025, 0.05, 0.1],
        "risk_coefficient": [0.1, 0.25, 0.5, 1.0],
        # "lambda_hhi": [0.0],
        # "lambda_turnover": [0.0],
        "lambda_turnover": [0.001, 0.003, 0.005, 0.01],
        "lambda_hhi": [0.03, 0.04, 0.05, 0.1],
        "seed": [random_seed, random_seed+11, random_seed+23],
    }
    raw = list(ParameterSampler(space, n_iter=n, random_state=random_seed))
    return [p for p in raw if (p["n_steps"] % p["batch_size"] == 0)] or raw


def eval_monthly_metrics(env, model, etf_list):
    """Evaluate over episode; collect turnover, deviation from EW, eff. N."""
    obs = env.reset()
    total_reward = 0.0

    N = len(etf_list)
    w_eq = np.ones(N) / N

    turnovers, devs, effns = [], [], []
    last_num_months = 0

    while True:
        action, _ = model.predict(obs, deterministic=True)
        obs, rewards, dones, infos = env.step(action)

        r = float(rewards[0]); info = infos[0]
        total_reward += r
        w = np.asarray(info.get("weights", np.ones(N)/N), dtype=float)
        if w.shape[0] != N or not np.all(np.isfinite(w)): w = np.ones(N)/N

        num_months = int(info.get("num_months", last_num_months))
        boundary = (abs(r) > 1e-12) or (num_months > last_num_months)
        if boundary:
            hhi = float(np.sum(np.square(w)))
            turnovers.append(float(info.get("last_turnover", 0.0)))
            devs.append(float(np.sum(np.abs(w - w_eq))))
            effns.append(1.0 / hhi if hhi > 1e-12 else np.nan)
            last_num_months = num_months

        if dones[0]:
            break

    return {
        "total_reward": total_reward,
        "avg_turnover": float(np.mean(turnovers)) if turnovers else 0.0,
        "avg_dev_from_ew": float(np.mean(devs)) if devs else 0.0,
        "avg_eff_n": float(np.nanmean(effns)) if effns else np.nan,
    }


def validate_and_tune(train_df: pd.DataFrame, val_df: pd.DataFrame,
                      etf_list: List[str], cfg: TrainingConfig,
                      random_seed: int) -> Dict[str, float]:

    candidates = sample_valid_params(cfg, random_seed, cfg.n_iter_tuning)

    # Encourage variation around EW but not too wild
    DEV_EW_MIN, DEV_EW_MAX = 0.25, 0.70
    TURNOVER_MAX = 0.60

    best_feasible = None   # (score, params, metrics)
    best_overall  = None

    for params in candidates:
        seed = params["seed"]
        risk_coeff = params["risk_coefficient"]
        lambda_turnover = params["lambda_turnover"]
        lambda_hhi = params["lambda_hhi"]

        set_global_seed(seed)

        # Train env
        env_train = make_env_from_frame(train_df, etf_list, cfg, risk_coeff,
                                        lambda_turnover, lambda_hhi,
                                        use_baseline=True, vecnorm_load_path=None, training=True)
        policy_kwargs = dict(net_arch=dict(pi=list(cfg.policy_arch), vf=list(cfg.policy_arch)))
        model = PPO(
            "MlpPolicy", env_train,
            device=cfg.device,
            learning_rate=params["learning_rate"],
            n_steps=params["n_steps"],
            batch_size=params["batch_size"],
            gamma=params["gamma"],
            gae_lambda=params["gae_lambda"],
            ent_coef=params["ent_coef"],
            clip_range=params["clip_range"],
            vf_coef=params["vf_coef"],
            max_grad_norm=params["max_grad_norm"],
            n_epochs=cfg.n_epochs,
            seed=seed,
            verbose=0,
            policy_kwargs=policy_kwargs
        )
        model.learn(total_timesteps=cfg.tuning_timesteps)

        # Val env (reuse stats in-memory)
        env_val = make_env_from_frame(val_df, etf_list, cfg, risk_coeff,
                                      lambda_turnover, lambda_hhi,
                                      use_baseline=True, vecnorm_load_path=None, training=False)
        copy_vecnorm_stats(env_train, env_val, cfg.use_vecnormalize)

        metrics = eval_monthly_metrics(env_val, model, etf_list)
        score = metrics["total_reward"]

        # Track best overall
        if (best_overall is None) or (score > best_overall[0]):
            best_overall = (score, params.copy(), metrics)

        # Feasible selection
        feasible = (DEV_EW_MIN <= metrics["avg_dev_from_ew"] <= DEV_EW_MAX) and (metrics["avg_turnover"] <= TURNOVER_MAX)
        if feasible and ((best_feasible is None) or (score > best_feasible[0])):
            best_feasible = (score, params.copy(), metrics)

    chosen = best_feasible if best_feasible is not None else best_overall
    assert chosen is not None, "No candidates evaluated."
    score, best_params, metrics = chosen

    print("\nTuning selection:")
    print(f"  score={score:.4f}")
    print(f"  λ_turnover={best_params['lambda_turnover']}, λ_hhi={best_params['lambda_hhi']}")
    print(f"  dev(EW)={metrics['avg_dev_from_ew']:.3f}, turnover={metrics['avg_turnover']:.3f}, effN={metrics['avg_eff_n']:.2f}")

    return best_params




# -------------------------
# Train with early stopping
# -------------------------
def train_with_early_stopping(train_df, val_df, etf_list, cfg, best_params, window_dir) -> Tuple[str, str]:
    seed = best_params["seed"]
    risk_coeff = best_params["risk_coefficient"]
    lambda_turnover = best_params["lambda_turnover"]
    lambda_hhi = best_params["lambda_hhi"]
    set_global_seed(seed)

    env_train = make_env_from_frame(train_df, etf_list, cfg, risk_coeff,
                                    lambda_turnover, lambda_hhi,
                                    use_baseline=True, vecnorm_load_path=None, training=True)

    policy_kwargs = dict(net_arch=dict(pi=list(cfg.policy_arch), vf=list(cfg.policy_arch)))
    model = PPO(
        "MlpPolicy", env_train,
        device=cfg.device,
        learning_rate=best_params["learning_rate"],
        n_steps=best_params["n_steps"],
        batch_size=best_params["batch_size"],
        gamma=best_params["gamma"],
        gae_lambda=best_params["gae_lambda"],
        ent_coef=best_params["ent_coef"],
        clip_range=best_params["clip_range"],
        vf_coef=best_params["vf_coef"],
        max_grad_norm=best_params["max_grad_norm"],
        n_epochs=cfg.n_epochs,
        seed=seed,
        verbose=0,
        policy_kwargs=policy_kwargs
    )

    # Build a single val env, copy stats each eval
    env_val = make_env_from_frame(val_df, etf_list, cfg, risk_coeff,
                                  lambda_turnover, lambda_hhi,
                                  use_baseline=True, vecnorm_load_path=None, training=False)

    model_path = os.path.join(window_dir, "best_ppo.zip")
    vecnorm_path = os.path.join(window_dir, "vecnormalize.pkl")  # persist for prediction if you prefer disk

    best_val_reward, no_improve = -np.inf, 0
    log_rows = []

    for step in range(0, cfg.max_timesteps, cfg.incremental_timesteps):
        model.learn(total_timesteps=cfg.incremental_timesteps)

        copy_vecnorm_stats(env_train, env_val, cfg.use_vecnormalize)
        obs = env_val.reset()
        val_reward = 0.0
        while True:
            action, _ = model.predict(obs, deterministic=True)
            obs, rewards, dones, infos = env_val.step(action)
            val_reward += float(rewards[0])
            if dones[0]:
                break

        log_rows.append({"training_step": step + cfg.incremental_timesteps,
                         "validation_reward": val_reward})

        if val_reward > best_val_reward:
            best_val_reward = val_reward
            no_improve = 0
            model.save(model_path)
            # If you prefer to keep VecNormalize on disk for pred:
            if cfg.use_vecnormalize:
                env_train.save(vecnorm_path)
        else:
            no_improve += 1
            if no_improve >= cfg.patience:
                break

    pd.DataFrame(log_rows).to_csv(os.path.join(window_dir, "training_validation_log.csv"), index=False)
    return model_path, vecnorm_path




# -------------------------
# Predict on pred_df & compute OOS wealth
# -------------------------
def predict_and_log(pred_df, etf_list, cfg, best_params, model_path, vecnorm_path, window_dir):
    seed = best_params["seed"]
    risk_coeff = best_params["risk_coefficient"]
    lambda_turnover = best_params["lambda_turnover"]
    lambda_hhi = best_params["lambda_hhi"]
    set_global_seed(seed)

    # Build pred env; if you persisted VecNormalize, load it here:
    if cfg.use_vecnormalize and os.path.exists(vecnorm_path):
        env_pred = make_env_from_frame(pred_df, etf_list, cfg, risk_coeff,
                                       lambda_turnover, lambda_hhi,
                                       use_baseline=True, vecnorm_load_path=vecnorm_path, training=False)
    else:
        env_pred = make_env_from_frame(pred_df, etf_list, cfg, risk_coeff,
                                       lambda_turnover, lambda_hhi,
                                       use_baseline=True, vecnorm_load_path=None, training=False)

    model = PPO.load(model_path, device=cfg.device)
    obs = env_pred.reset()

    dates, weights, wealth_series = [], [], []
    last_num_months = 0

    while True:
        action, _ = model.predict(obs, deterministic=True)
        obs, rewards, dones, infos = env_pred.step(action)

        r = float(rewards[0]); inf = infos[0]
        num_months = inf.get("num_months", last_num_months)
        boundary = (abs(r) > 1e-12) or (num_months > last_num_months)
        if boundary:
            dates.append(inf["date"])
            weights.append(inf["weights"])
            wealth_series.append(inf["cumulative_wealth"])
            last_num_months = num_months

        if dones[0]:
            break

    weights_df = pd.DataFrame(weights, columns=etf_list); weights_df.insert(0, "Date", dates)
    weights_df.to_csv(os.path.join(window_dir, "weights.csv"), index=False)

    wealth_df = pd.DataFrame({"Date": dates, "Cumulative_Wealth": wealth_series})
    wealth_df.to_csv(os.path.join(window_dir, "wealth.csv"), index=False)

    wealth_arr = np.array(wealth_series, dtype=float)
    month_rets = wealth_arr[1:] / wealth_arr[:-1] - 1.0 if len(wealth_arr) > 1 else np.array([])
    if month_rets.size > 0:
        mean = month_rets.mean()
        vol = month_rets.std(ddof=1)
        sharpe = mean / vol * np.sqrt(12) if vol > 0 else np.nan
        cum = (1.0 + month_rets).cumprod()
        peak = np.maximum.accumulate(cum)
        mdd = np.min(cum / peak - 1.0)
    else:
        mean = vol = sharpe = mdd = np.nan

    return float(wealth_arr[-1] - 1.0 if wealth_arr.size else 0.0), float(sharpe), float(mdd)
import os
import pandas as pd

def _unwrap_to_portfolio_env(venv):
    """
    Walk through VecNormalize -> VecEnv -> wrappers to reach the base PortfolioEnv.
    Safe even if some layers are missing.
    """
    env = venv
    # VecNormalize
    if hasattr(env, "venv"):
        env = env.venv
    # VecEnv (Dummy/Sub proc) -> take first
    if hasattr(env, "envs") and len(env.envs) > 0:
        env = env.envs[0]
    # Generic wrappers chain
    seen = set()
    while hasattr(env, "env") and id(env) not in seen:
        seen.add(id(env))
        env = env.env
    return env


def save_feature_list_once(feature_df, etf_list, cfg, out_dir, filename_base="obs_feature_columns", force_overwrite: bool=False):
    txt_path = os.path.join(out_dir, f"{filename_base}.txt")
    csv_path = os.path.join(out_dir, f"{filename_base}.csv")

    if (not force_overwrite) and os.path.exists(txt_path) and os.path.exists(csv_path):
        print(f"[info] Feature list already saved: {txt_path}")
        return


    # Build a light env ONCE (no training) using your standard builder
    probe_env = make_env_from_frame(
        feature_df, etf_list, cfg,
        risk_coeff=getattr(cfg, "default_risk_coeff", 0.05),
        lambda_turnover=getattr(cfg, "lambda_turnover", 0.0),
        lambda_hhi=getattr(cfg, "lambda_hhi", 0.0),
        use_baseline=True,
        vecnorm_load_path=None,
        training=False
    )

    base_env = _unwrap_to_portfolio_env(probe_env)
    feat_cols = list(base_env.feature_cols)
    
    # Safety assertions
    assert all(not c.startswith("Actual_Return_") for c in feat_cols), \
        "Leakage guard: Actual_Return_* should not be in observation features."
    if cfg.exclude_price_levels:
        assert all(not c.startswith("Price_") for c in feat_cols), \
            "Price_* should be excluded when exclude_price_levels=True."
    
    if not hasattr(base_env, "feature_cols"):
        probe_env.close()
        raise RuntimeError("Underlying env does not expose `feature_cols`.")

    feature_names = list(base_env.feature_cols)

    # Save once (TXT + CSV)
    os.makedirs(out_dir, exist_ok=True)
    with open(txt_path, "w", encoding="utf-8") as f:
        f.write("# Observation feature columns (one per line)\n")
        f.write(f"# lookback_period={getattr(base_env, 'lookback_period', getattr(cfg, 'lookback_period', 'NA'))}\n")
        f.write(f"# exclude_price_levels={getattr(base_env, 'exclude_price_levels', getattr(cfg, 'exclude_price_levels', 'NA'))}\n")
        for c in feature_names:
            f.write(f"{c}\n")

    pd.DataFrame({"feature": feature_names}).to_csv(csv_path, index=False)

    # Optional quick sanity check: len(features)*lookback == obs dim
    try:
        obs_dim = int(probe_env.observation_space.shape[0])
        lp = int(getattr(base_env, "lookback_period", getattr(cfg, "lookback_period", 1)))
        expected = len(feature_names) * lp
        if expected != obs_dim:
            print(f"[WARN] obs dim mismatch: len(features)*lookback={expected} vs obs_space={obs_dim}")
    except Exception:
        pass

    probe_env.close()
    print(f"[info] Saved observation feature list to:\n  - {txt_path}\n  - {csv_path}")


```

```python
import pandas as pd
import glob
import os

# Configuration
output_dir = 'stage2_iterations'
pattern = os.path.join(output_dir, 'iteration_*', 'window_*', 'weights.csv')

# Find all weight files
files = glob.glob(pattern)

# --- Pass 1: Determine the master ETF column order ---
etf_columns = None
for file_path in files:
    df_temp = pd.read_csv(file_path, nrows=1)
    possible_etfs = [c for c in df_temp.columns if c not in ['Date', 'Iteration', 'Window']]
    if etf_columns is None:
        etf_columns = possible_etfs
    else:
        etf_columns = sorted(set(etf_columns).union(possible_etfs))

# --- Pass 2: Read, align, and combine ---
all_weights = []
for file_path in files:
    parts = file_path.split(os.sep)
    iteration = int(parts[-3].split('_')[1]) + 1  # shift to start from 1
    window = int(parts[-2].split('_')[1])

    df = pd.read_csv(file_path, parse_dates=['Date'])

    # Ensure all ETF columns exist and are in the same order
    for col in etf_columns:
        if col not in df.columns:
            df[col] = 0.0
    df = df[['Date'] + etf_columns]

    # Add window and iteration (iteration will be moved later)
    df.insert(0, 'Window', window)
    df['Iteration'] = iteration  # temp position at end

    all_weights.append(df)

# Combine
combined_df = pd.concat(all_weights, ignore_index=True)
combined_df.sort_values(['Iteration', 'Window', 'Date'], inplace=True)

# Move Iteration to end
iteration_col = combined_df.pop('Iteration')
combined_df['Iteration'] = iteration_col

# Save
out_path = os.path.join(output_dir, 'combined_weights.csv')
combined_df.to_csv(out_path, index=False)

print(f"Combined weights saved to: {out_path}")
```
